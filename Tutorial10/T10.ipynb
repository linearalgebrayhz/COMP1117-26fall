### Task 5: skip the bad lines

- Blank lines: ignore, do not count as skipped.
- Lines int() rejects: skip and count them.
- Round the average to 1 decimal place.
- No valid numbers: the average is 0.0.

Sample input file: './files/steps.txt'

Code provided by the teaching team:


````python
def average_steps(filename):
    total = 0
    count = 0
    skipped = 0
    with open(filename) as fin:
        for line in fin:
            text = line.strip()
            if text == '':
                continue
            try:
                total += int(text)
                count += 1
            except ValueError:
                skipped += 1
    if count == 0:
        return 0.0, skipped
    return round(total / count, 1), skipped
````

````python
# An example of Python OOP, you can play around with it
import math

class complexNum:

    def __init__(self, re, im) -> None: # This is called constructor method
        # initialize
        self.re = re
        self.im = im

    def __add__(self, c): # This is called operator overloading
        return complexNum(self.re + c.re, self.im + c.re)
    
    def __neg__(self):
        return complexNum(-self.re, self.im)

    def __repr__(self) -> str: # this method determines what the object look like 
        return f"{self.re} + {self.im}i"
    
    def __eq__(self, c):
        return self.re == c.re and self.im == c.im
    
    @staticmethod # This is a method bind to the whole complex class
    def modulus(c):
        return math.sqrt(c.re ** 2 + c.im ** 2)

    
c1 = complexNum(1,1)

c2 = complexNum(1,1)

print(c1 == c2) 

````
