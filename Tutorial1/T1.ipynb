### Part 1 What Are Python and VS Code, and Why Do You Need Both?

When you "run a program", two different things are involved:

1. **Python** (the *translator*) - a program installed on your computer that reads Python code and executes it, line by line. It has no window of its own to type in; it just works in the background. Free, from [python.org](https://www.python.org/downloads/).
2. **VS Code** (the *notebook*) - an editor: a place to type code, save it as a `.py` file, and press one button to ask Python to run it. It colours your code, flags typos, and shows the output. By itself it **cannot** run Python - it calls the translator. Free, from [code.visualstudio.com](https://code.visualstudio.com/).

You need both: VS Code is where you *write*; Python is what *runs* what you wrote.

> "I installed VS Code, why can't I run my program?" → you have the notebook but no translator - install Python.
>
> "I installed Python, why is there no window to type in?" → Python works in the background; you still need an editor.

### Design first, then implement

Before typing any code, work out *what steps* solve the problem - in plain language or pseudocode. Only once the steps make sense do you translate them into Python. This is exactly what Task 2 asked of you: the formula (`°F = °C × 9/5 + 32`) is the *design*; `fahrenheit = ...` is the *implementation*.

### Three Kinds of Errors

1. **Syntax errors** — the code isn't valid Python at all (e.g. a missing `)` or a Chinese quote `“`). Python refuses to even start running it. This is what the "Common Problems Checklist" above mostly deals with.
2. **Runtime errors** (exceptions) — the code is valid Python, but something goes wrong *while* it runs (e.g. dividing by zero, casting `"abc"` to `int`). The program crashes partway through.
3. **Logic errors** — the code runs fine and prints *something*, but the answer is wrong (e.g. using `-` instead of `+` in the Fahrenheit formula). These are the hardest to catch, because Python gives you no error message at all.

`````python
# Runtime error: uncomment the next line to see Python stop with a traceback
# result = 10 / 0

# Logic error: this *runs* fine, but the formula is wrong
celsius = 0
wrong_fahrenheit = celsius * 9 / 5 - 32   # should be + 32, not - 32
print("Wrong:", wrong_fahrenheit)          # prints -32.0, not 32.0

correct_fahrenheit = celsius * 9 / 5 + 32
print("Correct:", correct_fahrenheit)
`````

### Check Claims by Running the Code

Never trust a formula (yours or an AI agent's) just because it *looks* right — **run it** on inputs where you already know the answer, like Task 2 asked (`0°C → 32°F`, `100°C → 212°F`). If the output matches, you've checked your claim; if not, you've caught a logic error before it reached anyone else.

This habit — testing against known answers — is exactly what later tutorials will formalise into writing real test cases (job #2 of this year's four tutorial jobs; see "Tutorial Arrangements" near the top of this notebook).
