# 2. *(extra)* Warm-up: Branches, Push & Pull

We did not have time for these last tutorial, and everything today builds on them. First **branches**, then **push and pull**.

## 2.1 `git branch` and `HEAD`

A branch is just a **name pointing at a commit** (a movable label). `HEAD` is Git's **"you are here"** marker: it points at the branch you are currently on. When you commit, the branch you are on moves forward and `HEAD` goes with it.

<img src='./assets/t5_head.png' width='880' alt='HEAD points at the branch you are on; branches are labels on commits'>

That is why branches are *cheap*: creating one only creates a new label (a few bytes), it never copies your files.

### Branch commands

| I want to... | Command | Notes |
|---|---|---|
| **List** local branches | `git branch` | `*` marks the current branch |
| List with last commit + tracking info | `git branch -vv` | shows `[origin/<name>]` if linked to GitHub |
| List local **and** remote branches | `git branch -a` | remote ones appear as `remotes/origin/...` |
| **Create** a branch, stay where I am | `git branch <name>` | |
| **Create and switch** | `git switch -c <name>` | older: `git checkout -b <name>` |
| **Switch** to an existing branch | `git switch <name>` | older: `git checkout <name>` |
| **Rename** the current branch | `git branch -m <new-name>` | |
| **Delete** a branch that is merged | `git branch -d <name>` | refuses if it holds unmerged work (a safety net) |
| **Force-delete** (throw the work away!) | `git branch -D <name>` | careful: those commits become hard to find |
| **Draw** the history of all branches | `git log --oneline --graph --all --decorate` | the best "where am I?" picture |

Rules to remember:

- You cannot delete the branch you are standing on - switch away first.
- Branch names: no spaces. Use hyphens or slashes, e.g. `feature-add-greeting`, `docs/update-readme`.
- Uncommitted edits **travel with you** when you switch (if they do not clash). Commit before switching if you want a clean separation.
- A new branch starts from **wherever you are now**, so `git switch main` + `git pull` first!

## 2.2 Push & Pull: keeping your laptop and GitHub in sync

You used `git push` and `git pull` on `main` in Tutorial 4. They behave the same on **every branch**, and all teamwork on GitHub depends on them.

You now have **two copies** of the history: the **local** repository on your laptop and the **remote** one on GitHub (`origin`). Push and pull are the two roads between them.

<img src='./assets/t5_push_pull.png' width='880' alt='git push uploads, git fetch downloads only, git pull is fetch plus merge'>

| Command | Direction | What it does | VS Code button |
|---|---|---|---|
| `git push` | laptop -> GitHub | Uploads **commits** you have and GitHub does not | **Publish Branch** (first time), **Sync Changes** / **Push** |
| `git fetch` | GitHub -> laptop | Downloads new commits but **does not touch your files** (safe, just looking) | *(runs automatically in the background)* |
| `git pull` | GitHub -> laptop | `fetch` **+** `merge`: downloads new commits **and** updates your current branch | **Pull**, **Sync Changes** |

Things to remember:

- Only **committed** work is pushed. Uncommitted edits stay on your laptop.
- `git push -u origin <branch>` (what **Publish Branch** does) links your local branch to the GitHub one, so later a plain `git push` / `git pull` knows where to go. `-u` is only needed **once per branch**.
- `git pull` updates the branch you are **currently on** - check your branch first!
- Git remembers where GitHub *was* the last time you talked to it, as **remote-tracking branches** such as `origin/main`. That is how it can say you are **ahead** or **behind**:

| `git status -sb` says | Meaning | What to do |
|---|---|---|
| `## main...origin/main` | In sync | nothing |
| `## main...origin/main [ahead 1]` | You have 1 commit GitHub lacks | **push** |
| `## main...origin/main [behind 1]` | GitHub has 1 commit you lack | **pull** |
| `## main...origin/main [ahead 1, behind 1]` | Both sides moved (diverged) | **pull** first (may merge), then **push** |

**Rule of thumb:** *pull before you start working, pull before you push, push when you finish a chunk of work.*

<img src='./assets/t5_ahead_behind.png' width='900' alt='Four states: in sync, ahead 1, behind 1, and diverged'>

## 2.3 Pushing a **new** branch: why `-u`?

A brand-new local branch does not exist on GitHub, and Git does not know where it should go. A plain `git push` therefore fails the first time, with *"The current branch has no upstream branch"*.

Git's error message even tells you the fix (`git push --set-upstream origin <branch>`). `-u` (**--set-upstream**) pushes the branch **and** remembers the link between your local branch and the matching branch on GitHub (`origin/<branch>`). This is exactly what VS Code's **Publish Branch** button does. After that, plain `git push` and `git pull` just work on this branch.

# 3. Collaboration Models

## 3.1 What is a Branch? *(Slide 5)*

**Concept:** an independent line of development extending from your repository history.

<img src='./assets/t5_branch_basic.png' width='920' alt='Commit graph: a feature branch splits from main and merges back'>

**Why branching matters**

- **Isolation** - develop new features without breaking working code on `main`.
- **Parallel work** - multiple team members work on separate tasks at the same time.
- **Safe experimentation** - test changes freely before merging into production.

Each `o` is a **commit**: a saved snapshot. A branch is just a movable label pointing at one commit, which is why creating a branch is instant and cheap. *(You met this in Section 2.1 - treat this slide as the quick version.)*

## 3.2 Important Terms *(Slide 6)*

| Term | Meaning |
|---|---|
| **`main`** | The primary, stable default branch of a repository |
| **Feature branch** | A temporary branch dedicated to one specific task or feature |
| **Switch / Checkout** | Moving your working environment from one branch to another |
| **Merge** | Integrating history and changes from one branch into another |
| **Pull Request (PR)** | A GitHub request to review and merge branch changes |

Two more words you will hear today:

| Term | Meaning |
|---|---|
| **`origin`** | The default nickname for the remote (GitHub) repository you cloned from |
| **Remote branch** | A branch that lives on GitHub, shown locally as `origin/<name>` |

## 3.3 Clone vs. Branch *(Slide 7)*

| | **Clone** | **Branch** |
|---|---|---|
| **What it does** | Downloads a *local copy* of a remote repo to your computer | Creates an *isolated line of work inside the same repo* |
| **Direction** | Remote GitHub -> Local PC | Inside the same repo |
| **Where** | On your laptop | Local, then published to GitHub |
| **Typical use** | Getting the project onto your laptop (once) | Each feature or bug fix, so teammates can work in parallel |

<img src='./assets/t5_clone_branch.png' width='960' alt='Clone copies the GitHub repository to your laptop; a branch is a parallel line of work inside one repo'>

> **Key point:** *clone* copies a repository **between places** (GitHub -> your laptop). A *branch* is a parallel line of work **inside** one repository. They are not alternatives: you **clone once**, then create **a branch for every task**.


## 3.4 The Team Workflow: In-Repo Branching *(Slide 8)*

**In-repo branching model (shared repository)**

- Used when **all team members have direct write access**.
- Everyone creates feature branches directly inside the group repository.
- Best for: **group projects and private team repos**.

<img src='./assets/t5_workflow_inrepo.png' width='840' alt='In-repo branching: Alice and Bob push feature branches to the same GitHub repo'>

# 5. Pull Requests

## 5.1 Pull Request (PR) & Branch Protection *(Slide 15)*

**Definition:** a GitHub feature for **proposing and discussing code changes before incorporating them into `main`**.

**Key benefits**

- Enables **peer review** and discussion on code quality.
- Displays **inline visual diffs** showing added and deleted lines.
- Allows **automatic testing checks** before merging.

**Branch protection rules**

- Real-world projects **block direct pushes to `main`** to prevent accidental breaking changes.
- All updates must go through a **feature branch and an approved Pull Request**.

<img src='./assets/t5_pr_flow.png' width='920' alt='Pull request flow from feature branch through review to main; direct pushes to main are blocked'>

> A PR is *not* a Git feature - it lives on GitHub. Under the hood it is simply "please run `git merge` for me, after people have looked at it".

## 5.2 Create a PR on GitHub *(Slide 16)*

1. Open your repository on **github.com**.
2. Click the yellow **"Compare & pull request"** banner for `feature-add-greeting`.
3. Fill in a **descriptive title and summary** (e.g. *Add personalized greeting function to README.md*).
4. Click **Create pull request**.
5. Click **Merge pull request** -> **Confirm merge**.

**AI tip:** use Copilot / GenAI to **draft PR summaries**, but **verify accuracy before publishing!**

**What reviewers see:** the *diff* between `main` and your branch. `git diff main...feature-add-greeting` prints the same thing as the "Files changed" tab (green `+` lines are additions):

### What "Merge pull request" does

Clicking the green button makes GitHub merge the branch into `main` **on the server**.

Back on your laptop, `main` has not moved yet. This is why **"Pull before you branch"** matters: GitHub changed, your laptop did not.

# 6. Merge Conflicts

## 6.1 What is a Merge Conflict? *(Slide 17)*

**Scenario:** two branches edit **the same lines of the same file differently** before merging. Git cannot guess which version you want, so it stops and asks *you*.

<img src='./assets/t5_conflict.png' width='920' alt='Two branches change the same line, so merging them conflicts'>

**Conflict markers** that Git writes into the file:

<img src='./assets/t5_conflict_markers.png' width='920' alt='Conflict markers: current change, separator, incoming change, and the VS Code accept buttons'>

**Resolution actions** in VS Code: **Accept Current Change** | **Accept Incoming Change** | **Accept Both Changes**

| Term | Which branch? |
|---|---|
| **Current** (`HEAD`) | The branch you are **on** - the one *receiving* the merge |
| **Incoming** | The branch you asked to **merge in** |

Conflict markers are just text. Git does not care which option you pick, but **the file must not contain `<<<<<<<`, `=======` or `>>>>>>>` when you finish**.

## 6.3 Troubleshooting Checklist *(Slide 21)*

| You see... | Likely cause | How to fix |
|---|---|---|
| **Permission denied (403)** when pushing | Trying to push directly to a repository where you **lack write access** | Ask the owner to add you as a **collaborator**, or push to a repository you own |
| **Push rejected / non-fast-forward** | The remote branch has **newer commits** than your local laptop | Click **Sync Changes** / **Pull** first to incorporate remote updates, then push |
| **Accidentally edited code on `main`** | Forgot to switch to a feature branch before typing | Click `main` in the status bar -> **+ Create Branch** (VS Code safely carries your uncommitted edits over!) |
| **`<<<<<<<` markers inside a file** | Unresolved merge conflict | Open file in VS Code, click **Accept Current / Incoming**, stage (+), and commit |

> **Habit:** read the **last few lines** of the error message. Git usually tells you exactly what to do next.

# 7. Best Practices, Cheat Sheet & Exit Check

## 7.1 Good Practices for Team Collaboration *(Slide 22)*

- **Keep `main` clean** - never code directly on `main`; always use feature branches.
- **Pull before branching** - run pull on `main` before creating a new branch, so you start from the latest code.
- **Meaningful branch names** - use clear prefixes like `feature/add-greeting`, `bugfix/fix-greeting-bug`, or `docs/update-readme`.
- **Clean up** - delete feature branches on GitHub after merging.

> **Naming note:** the exercises use `feature-add-greeting` (hyphens) while the best-practice slide shows `feature/add-greeting` (slash). Both are valid. Slashes group branches into folders in tools like VS Code and GitHub, and the prefix tells teammates the *purpose* of the branch at a glance. Pick one convention per team and stick to it.

If not yet, finish Task 1 after the tutorial. Try the quick quiz below to check yourself:

1. What is the difference between *cloning* and *branching*?
2. You are on `main` and have already typed some code. What is the safest next move?
3. What does **Publish Branch** do, and what does it *not* do?
4. During a merge conflict, which branch is **Current** and which is **Incoming**?
5. Git says your push was **rejected (non-fast-forward)**. What do you do?
6. The PR was merged. What should you do with the feature branch?
7. A new branch's first plain `git push` fails with *no upstream branch*. Which command fixes it, and what does `-u` do?
8. What is the difference between `git fetch` and `git pull`? If `git status -sb` shows `[ahead 2]`, which one do you need?

<details>
<summary><b>Answers (click to reveal)</b></summary>

1. **Clone** copies the repository from GitHub to your laptop (once). A **branch** is a separate line of work inside the repository (one per task).
2. Click `main` in the status bar -> **+ Create new branch...**. Your uncommitted edits move to the new branch and `main` stays clean.
3. It uploads (`git push -u origin <branch>`) your local branch and its commits to GitHub. It does **not** merge anything into `main` - that needs a Pull Request.
4. **Current** = the branch you are standing on (the one receiving the merge). **Incoming** = the branch you are merging in.
5. **Sync Changes** (pull, then push). Remote has commits you do not; bring them in first.
6. **Delete** it on GitHub (and locally). Then pull `main` before starting the next feature branch.
7. `git push -u origin <branch>`. `-u` links your local branch to the GitHub one, so later `git push` / `git pull` know where to go (VS Code's **Publish Branch** does this).
8. `fetch` only downloads new commits (your files do not change); `pull` = `fetch` + `merge`, so it also updates your current branch. `[ahead 2]` means you have 2 commits GitHub lacks, so you need **`git push`**.
</details>
