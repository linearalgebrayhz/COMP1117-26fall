### 1.1 Sets

A set is an **unordered** collection with **no duplicate elements**.

- Create: `s = {1, 2, 3}` or `s = set()` (note: `{}` is a *dict*, not a set!)
- Add / remove: `s.add(x)`, `s.remove(x)` (error if missing), `s.discard(x)` (no error)
- Membership: `x in s` - this is **fast** (O(1) on average), unlike `in` on a list (O(n))

````python
# Sets: creation and basic operations
fruits = {'apple', 'banana', 'cherry'}
print(fruits)

fruits.add('durian')
fruits.add('apple')      # adding a duplicate does nothing
print(fruits)

fruits.discard('banana')
print('apple' in fruits)  # membership test: very fast
print('mango' in fruits)

# Creating a set from a list removes duplicates automatically
nums = [1, 2, 2, 3, 3, 3, 4]
print(set(nums))
````

### 1.2 Set operations

| Operation | Operator | Method |
|---|---|---|
| Union | `A \| B` | `A.union(B)` |
| Intersection | `A & B` | `A.intersection(B)` |
| Difference | `A - B` | `A.difference(B)` |
| Symmetric difference | `A ^ B` | `A.symmetric_difference(B)` |
| Subset test | `A <= B` | `A.issubset(B)` |

````python
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7}

print('union:      ', A | B)   # in either set
print('intersect:  ', A & B)   # in both
print('difference: ', A - B)   # in A but not B
print('sym diff:   ', A ^ B)   # in exactly one of them
print('{1,2} <= A? ', {1, 2} <= A)  # subset
````

### 1.3 Dictionaries

A dictionary maps **keys** to **values**. Keys must be unique and hashable (strings, numbers, tuples - but **not lists**).

- Create: `d = {'a': 1, 'b': 2}` or `d = {}`
- Access: `d['a']` (error if missing) or `d.get('a', default)` (safe)
- Insert/update: `d['c'] = 3`
- Delete: `del d['a']` or `d.pop('a')`
- Loop: `for k, v in d.items(): ...`

Since Python 3.7, dictionaries remember **insertion order**.

````python
scores = {'Alice': 90, 'Bob': 75}

scores['Carol'] = 82          # insert
scores['Bob'] = 80            # update (keys are unique!)

print(scores['Alice'])        # direct access
print(scores.get('Dave', 0))  # safe access with default

print(list(scores.keys()))
print(list(scores.values()))

for name, score in scores.items():
    print(f'{name}: {score}')

print('Alice' in scores)      # 'in' checks KEYS, not values
````

### 1.5 More dict methods (from yesterday's lecture)

- `d.update(other)` - merge another dict into `d` (later keys win)
- `d.pop(key, default)` - remove and return the value (no error with a default)
- `d.setdefault(key, default)` - get the value, inserting the default if the key is missing
- `d.keys()`, `d.values()`, `d.items()` - *views* of the dict (live: they change when the dict changes)
- `sorted(d)` - iterates keys in sorted order; `sorted(d.items(), key=lambda kv: kv[1])` sorts by value
- Nested data is common: a dict of dicts, or a dict whose values are lists

````python
# Dict methods in action
menu = {'coffee': 25, 'tea': 20}
menu.update({'tea': 22, 'cake': 35})   # update tea, add cake
print(menu)

print(menu.pop('coffee', 0))           # remove + return
print(menu.pop('juice', 0))            # no KeyError thanks to the default

menu.setdefault('tea', 99)             # 'tea' exists -> no change
menu.setdefault('scone', 18)           # missing -> inserted
print(menu)

# Nested structures: a gradebook (dict of dicts)
grades = {'amy': {'midterm': 80, 'final': 85},
          'ben': {'midterm': 72, 'final': 90}}
print(grades['amy']['final'])          # index twice to go two levels deep

for student, marks in grades.items():
    avg = (marks['midterm'] + marks['final']) / 2
    print(f'{student}: average {avg}')
````

### 1.6 Common pitfalls (good exam traps!)
1. `{}` creates an empty **dict**, not a set. Use `set()`.
2. Sets are **unordered** - don't rely on print order, and no indexing (`s[0]` is an error).
3. Dict keys must be **hashable**: `d[[1,2]] = 'x'` raises `TypeError`. Use a tuple instead.
4. `x in d` checks **keys**. To check values: `x in d.values()`.
5. Don't add/remove keys from a dict **while looping over it** - loop over `list(d)` instead.

## 2. Comprehensions - revision

Yesterday's lecture ended with **comprehensions**: compact one-liners that build a list, set or dict from a loop. They do the same job as a `for` loop with `.append()` - just shorter and (usually) faster.

### 2.1 List comprehension

```python
[expression  for x in iterable  if condition]
```

is exactly the same as:

```python
result = []
for x in iterable:
    if condition:
        result.append(expression)
```

````python
# Loop version vs comprehension version - same result
squares = []
for n in range(10):
    squares.append(n ** 2)
print(squares)

print([n ** 2 for n in range(10)])          # map: transform every item
print([n for n in range(20) if n % 3 == 0]) # filter: keep some items
print([n ** 2 for n in range(20) if n % 3 == 0])  # both!

# The expression can be anything - even a function call
words = ['Hello', 'WORLD', 'PyThOn']
print([w.lower() for w in words])
print([len(w) for w in words])
````

### 2.2 Set comprehension

Same idea, with curly braces: `{expression for x in iterable}` - duplicates collapse automatically.

````python
# Set comprehension: unique results for free
nums = [1, 2, 2, 3, 3, 3, 4]
print({n % 3 for n in nums})          # distinct remainders

sentence = 'the cat sat on the mat'
print({len(w) for w in sentence.split()})   # distinct word lengths

# Classic exam one-liner: letters that appear in both strings
s1, s2 = 'hello', 'world'
print({ch for ch in s1 if ch in s2})
````

### 2.3 Dict comprehension

`{key_expr: value_expr for x in iterable}` - this is where comprehensions really pay off.

````python
# Dict comprehension
words = ['apple', 'banana', 'cherry']
print({w: len(w) for w in words})               # word -> length

print({n: n ** 2 for n in range(6)})            # n -> n squared

# Swap keys and values (the invert one-liner from exercise 5)
d = {'a': 1, 'b': 2}
print({v: k for k, v in d.items()})

# Count word frequencies with a comprehension + count()
sentence = 'the cat sat on the mat the dog sat'
ws = sentence.split()
freq = {w: ws.count(w) for w in set(ws)}        # note: fine for short lists
print(freq)
````

### 2.4 Pitfalls and when NOT to use a comprehension

1. Brackets decide the type: `[]` list, `{}` with `:` dict, `{}` without `:` set. `(x for x in ...)` is a **generator**, not a tuple!
2. If the body needs more than one or two lines, use a normal loop - readability wins.
3. `{w: ws.count(w) for w in ws}` recounts duplicates - put a `set(...)` around the iterable (or use the `d.get` loop from exercise 3, which is O(n)).

````python
# Generator trap: parentheses do NOT make a tuple comprehension
g = (n ** 2 for n in range(5))
print(g)                # a generator object - computed lazily
print(list(g))          # force it into a list
print(tuple(n ** 2 for n in range(5)))   # wrap in tuple() instead
````

### 1. Unique characters (warm-up)

Given a string, count how many **distinct** characters it contains (ignore case and spaces).

```
"Hello World" -> 8   # {h, e, l, o, w, r, d} ... count them!
```

Then: given two strings, find the characters that appear in **both**.

````python
def count_unique(s):
    """Return the number of distinct characters in s (case-insensitive, ignoring spaces)."""
    chars = set()
    for ch in s.lower():
        if ch != ' ':
            chars.add(ch)
    return len(chars)

print(count_unique('Hello World'))   # 7: h,e,l,o,w,r,d

s1, s2 = 'hello', 'world'
print(set(s1) & set(s2))   # {'l', 'o'} - letters in both
print(set(s1) - set(s2))   # letters only in s1
````

### 2. Removing duplicates while keeping order

`set(nums)` removes duplicates but loses the order. Write a function that removes duplicates from a list **while preserving the first-seen order**.

```
[3, 1, 2, 3, 2, 1] -> [3, 1, 2]
```

Hint: `dict.fromkeys(lst)` builds a dict whose keys are the list items (duplicates collapse, order kept).

````python
def dedup(lst):
    """Remove duplicates, preserving first-seen order."""
    return list(dict.fromkeys(lst))

print(dedup([3, 1, 2, 3, 2, 1]))        # [3, 1, 2]

# The "manual" version - same idea, shows what fromkeys does for you:
def dedup_manual(lst):
    seen = set()
    result = []
    for x in lst:
        if x not in seen:   # O(1) lookup - this is why sets beat lists here
            seen.add(x)
            result.append(x)
    return result

print(dedup_manual([3, 1, 2, 3, 2, 1])) # [3, 1, 2]
````

### 3. Word frequency counter

The classic dict exercise. Given a sentence, count how many times each word appears, then print the words from **most frequent to least frequent**.

```
"the cat sat on the mat the dog sat" -> the: 3, sat: 2, cat: 1, on: 1, mat: 1, dog: 1
```

Remember `lambda` from Tutorial 6? Use it in `sorted(..., key=...)`.

````python
sentence = 'the cat sat on the mat the dog sat'

freq = {}
for word in sentence.split():
    freq[word] = freq.get(word, 0) + 1   # get with default: no KeyError

print(freq)

# sort by count, highest first (lambda from Tutorial 6!)
for word, count in sorted(freq.items(), key=lambda kv: kv[1], reverse=True):
    print(f'{word}: {count}')
````

### 4. Student enrollment (set operations)

Two courses share some students:

```
comp1117 = {'amy', 'ben', 'chris', 'dora', 'eli'}
math1011 = {'ben', 'dora', 'fay', 'gus'}
```

Find:
1. Students taking **both** courses
2. Students taking **only COMP1117**
3. Students taking **exactly one** of the two courses
4. Is `{'amy', 'eli'}` a subset of COMP1117?

````python
comp1117 = {'amy', 'ben', 'chris', 'dora', 'eli'}
math1011 = {'ben', 'dora', 'fay', 'gus'}

print('both:          ', comp1117 & math1011)
print('only comp1117: ', comp1117 - math1011)
print('exactly one:   ', comp1117 ^ math1011)
print('either course: ', comp1117 | math1011)
print('subset?        ', {'amy', 'eli'} <= comp1117)

# Bonus: how many distinct students in total?
print('total distinct:', len(comp1117 | math1011))
````

### 5. Inverting a dictionary

Write a function that swaps keys and values:

```
{'a': 1, 'b': 2} -> {1: 'a', 2: 'b'}
```

**Follow-up:** what if two keys map to the same value? 
`{'a': 1, 'b': 1, 'c': 2}` - one value must map to a **list** of keys: `{1: ['a', 'b'], 2: ['c']}`

````python
def invert(d):
    """Swap keys and values. Assumes values are unique."""
    inv = {}
    for k, v in d.items():
        inv[v] = k
    return inv

print(invert({'a': 1, 'b': 2}))   # {1: 'a', 2: 'b'}

def invert_grouped(d):
    """Swap keys and values, grouping keys that share a value into a list."""
    inv = {}
    for k, v in d.items():
        if v not in inv:
            inv[v] = []
        inv[v].append(k)
    return inv

print(invert_grouped({'a': 1, 'b': 1, 'c': 2}))   # {1: ['a', 'b'], 2: ['c']}

# one-liner fans: invert(d) is also {v: k for k, v in d.items()}
print({v: k for k, v in {'a': 1, 'b': 2}.items()})
````

### 6. Two Sum (challenge)

Given a list of integers and a target, find the **indices** of two numbers that add up to the target.

```
nums = [2, 7, 11, 15], target = 9 -> (0, 1)   # nums[0] + nums[1] = 2 + 7 = 9
```

First write the $O(n^2)$ brute force (check every pair). Then use a dict to get $O(n)$: as you scan, store each number's index; for each `x`, check if `target - x` was already seen. 

This is a classic interview question - and a great demo of *why* dict lookup speed matters.

````python
def two_sum_brute(nums, target):
    """O(n^2): check every pair."""
    n = len(nums)
    for i in range(n):
        for j in range(i + 1, n):
            if nums[i] + nums[j] == target:
                return (i, j)
    return None

def two_sum(nums, target):
    """O(n): dict lookup for the complement."""
    seen = {}   # number -> index
    for i, x in enumerate(nums):
        if target - x in seen:
            return (seen[target - x], i)
        seen[x] = i
    return None

nums, target = [2, 7, 11, 15], 9
print(two_sum_brute(nums, target))   # (0, 1)
print(two_sum(nums, target))         # (0, 1)

import time
big = list(range(6000)) + [10000]
start = time.time(); two_sum(big, 11999); mid = time.time()
two_sum_brute(big, 11999); end = time.time()
print(f'dict version:   {mid-start:.4f}s')
print(f'brute force:    {end-mid:.4f}s')
````

### 7. Group anagrams (optional, if time permits)

Words that are anagrams of each other (same letters, rearranged) should be grouped together:

```
['eat', 'tea', 'tan', 'ate', 'nat', 'bat']
-> [['eat', 'tea', 'ate'], ['tan', 'nat'], ['bat']]
```

Key insight: sorted letters of anagrams are identical - `sorted('eat') == sorted('tea') == ['a','e','t']`. Use that as a dict **key** (a tuple, since lists can't be keys!).

````python
def group_anagrams(words):
    groups = {}
    for w in words:
        key = tuple(sorted(w))   # hashable key: ('a','e','t') for eat/tea/ate
        if key not in groups:
            groups[key] = []
        groups[key].append(w)
    return list(groups.values())

words = ['eat', 'tea', 'tan', 'ate', 'nat', 'bat']
print(group_anagrams(words))
````

## 4. Key takeaways
1. Use a **set** when you only care about *membership* and *uniqueness* - `in` is O(1).
2. Use a **dict** when you need to *associate* one thing with another (counts, indices, mappings).
3. `d.get(key, default)` avoids KeyError - the standard counting pattern is `d[w] = d.get(w, 0) + 1`.
4. Keys must be hashable: strings, numbers, tuples OK; lists are not.
5. The `| & - ^` operators make set questions almost one-liners - expect them on the midterm.
6. Reach for a **comprehension** when the loop body is a single expression; reach for a loop when it is not.
